```python
import pandas as pd
import numpy as np
import os

# Load the original dataset
file_path = "food_orders.csv"

raw_df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Total Records:", len(raw_df))
print("Total Columns:", len(raw_df.columns))
```

```python
# Batch loading into MySQL

import pandas as pd
import numpy as np

batch_size = 100

# Map CSV columns to MySQL columns
columns = [
    "order_id",
    "customer_id",
    "restaurant_id",
    "order_datetime",
    "delivery_datetime",
    "order_value",
    "delivery_fee",
    "payment_method",
    "discounts_and_offers",
    "commission_fee",
    "payment_processing_fee",
    "refunds_chargebacks",
    "weather_condition",
    "order_channel"
]

# Check that the MySQL table is empty
cursor = connection.cursor()
cursor.execute("SELECT COUNT(*) FROM raw_food_orders")
existing_records = cursor.fetchone()[0]
cursor.close()

if existing_records != 0:
    raise ValueError(
        "Table is not empty. Check existing records before inserting."
    )

# Convert missing values to SQL NULL
data_df = raw_df.astype(object).where(
    pd.notna(raw_df), None
)

# Prepare SQL insertion query
placeholders = ", ".join(["%s"] * len(columns))

insert_query = f"""
INSERT INTO raw_food_orders ({', '.join(columns)})
VALUES ({placeholders})
"""

total_loaded = 0
batch_number = 1

cursor = connection.cursor()

try:
    for start in range(0, len(data_df), batch_size):

        end = min(start + batch_size, len(data_df))

        batch_df = data_df.iloc[start:end]

        records = []

        for row in batch_df.itertuples(index=False, name=None):
            records.append(tuple(
                value.item() if isinstance(value, np.generic)
                else value
                for value in row
            ))

        try:
            cursor.executemany(insert_query, records)
            connection.commit()

            total_loaded += len(records)

            print(
                f"Batch {batch_number}: "
                f"{len(records)} records loaded successfully"
            )

        except Exception:
            connection.rollback()
            raise

        batch_number += 1

finally:
    cursor.close()

print("\nBatch loading completed!")
print("Total records loaded:", total_loaded)
print("Total batches:", batch_number - 1)
```

```python
# Task 3: Retrieve raw data from MySQL

import pandas as pd

# Read records from MySQL in batches of 100

batch_size = 100
offset = 0
batch_number = 1

retrieved_batches = []

while True:

    query = f"""
    SELECT *
    FROM raw_food_orders
    LIMIT {batch_size} OFFSET {offset}
    """

    batch_df = pd.read_sql(query, connection)

    if batch_df.empty:
        break

    retrieved_batches.append(batch_df)

    print(
        f"Batch {batch_number}: "
        f"{len(batch_df)} records retrieved"
    )

    offset += batch_size
    batch_number += 1

print("\nRaw data retrieval completed!")
print("Total records retrieved:",
      sum(len(batch) for batch in retrieved_batches))
```

```python
# Document rejection reasons

def get_rejection_reason(row):

    reasons = []

    if pd.isna(row["order_datetime"]):
        reasons.append("Missing or invalid order timestamp")

    if pd.isna(row["delivery_datetime"]):
        reasons.append("Missing or invalid delivery timestamp")

    if pd.notna(row["delivery_time_minutes"]):

        if row["delivery_time_minutes"] <= 0:
            reasons.append("Non-positive delivery duration")

    return "; ".join(reasons)


rejected_df["rejection_reason"] = rejected_df.apply(
    get_rejection_reason,
    axis=1
)

print("Rejected Records:")
display(rejected_df)
```

```python
import pandas as pd
import numpy as np

# Convert missing values to None and prepare Python-compatible values
def convert_value(value):
    if pd.isna(value):
        return None
    if isinstance(value, pd.Timestamp):
        return value.to_pydatetime()
    if isinstance(value, np.generic):
        return value.item()
    return value

records = [
    tuple(convert_value(value) for value in row)
    for row in cleaned_df.itertuples(index=False, name=None)
]

insert_query = """
INSERT INTO food_orders_analytics_db.cleaned_food_orders (
    order_id, customer_id, restaurant_id,
    order_datetime, delivery_datetime,
    order_value, delivery_fee, payment_method,
    discounts_and_offers, commission_fee,
    payment_processing_fee, refunds_chargebacks,
    weather_condition, order_channel,
    delivery_time_minutes, order_date,
    order_hour, platform_net_revenue
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s,
        %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

cursor = connection.cursor()

batch_size = 100
successful_batches = 0

try:
    for start in range(0, len(records), batch_size):
        batch = records[start:start + batch_size]

        cursor.executemany(insert_query, batch)
        connection.commit()

        successful_batches += 1
        print(f"Batch {successful_batches} inserted: {len(batch)} records")

    print("Total inserted:", len(records))

except Exception as e:
    connection.rollback()
    print("Insertion error:", e)

finally:
    cursor.close()
```
